```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Membaca dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

df = pd.read_csv(url, names=columns)

print(df.head())

# Untuk mencari missing values yang terdapat pada data, dan jumlah duplikasinya
print("Missing values:\n", df.isna().sum())
print("Jumlah duplikasi:", df.duplicated().sum())


# Untuk mengetahui statistik dasar data set mulai(mean, median, standard deviation, skewness dan kurtosis)
numeric_df = df.select_dtypes(include="number")

statistics = pd.DataFrame({
    "Mean": numeric_df.mean(),
    "Median": numeric_df.median(),
    "Standard deviation": numeric_df.std(),
    "Skewness" : numeric_df.skew(),
    "Kurtosis" : numeric_df.kurtosis(),

})

print(statistics)

#Melakukan visualisasi dengan histogram
df.hist(bins=15, figsize=(10, 6), edgecolor='black')
plt.suptitle('Histogram Distribusi Variabel Numerik', fontsize=14)
plt.show()

#Boxplot untuk mendeteksi outliers
plt.figure(figsize=(8, 5))
sns.boxplot(data=numeric_df, palette='Set2')
plt.title('Boxplot Mendeteksi Outliers')
plt.show()

#Scatter Plot untuk melihat korelasi
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='petal_length',
    y='petal_width',
    hue='species',
    palette='deep',
    s=70,
)
plt.title('Scatter Plot: Petal Length vs Petal Width')
plt.xlabel('Petal Length')
plt.ylabel('Petal Width')
plt.legend(title='Species')
plt.show()

# Heatmap korelasi
korelasi = numeric_df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    korelasi,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0
)

plt.title("Heatmap Korelasi Pearson")
plt.show()

```
